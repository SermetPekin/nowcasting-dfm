# Dissecting the Model: Structural Block Contributions

In a Mixed-Frequency Dynamic Factor Model, the target variable (like GDP) is not generated directly. Instead, the model maps dozens of macroeconomic metrics to separate, hidden **Latent Factors** defined by your logical "Blocks":

1. **Global Factor:** The overall baseline vector moving all macroeconomic data combined.
2. **Soft Factor:** The isolated noise derived specifically from Sentiment surveys and PMIs.
3. **Real Factor:** Hard activity metrics like Industrial Production and Retail Sales.

By slicing the Kalman Smoother's `Lambda` matrix ($\Lambda$), we can decompose the final Nowcast prediction mathematically back into these independent block trajectories to identify what sector of the economy is dragging the model up or down at any given time.

```python
from dfm_sp import Options
from dfm_sp import run_with_options, run
from dfm_sp import plot_block_contributions

# 1. Run the baseline New York Fed example
options = Options(
    vintage='2016-12-16',
    use_cache=True
)

Spec, X_old, Time_old, _ = run_with_options(options)

# `run()` mathematically processes the matrix yielding the Kalman outputs
ResObject = run(X_old, Spec, options)
```

### Plotting the Decomposition

The `plot_block_contributions` function automatically isolates the $\Lambda_{GDP} \times F_{Factor}$ multiplication map for a specific series. 

Below, observe how the Demeaned Signal (black line) behaves simply as the stacked structural aggregate of the Block models.

```python
# Decompose Real GDP (GDPC1) into its Soft vs Real components
fig = plot_block_contributions(ResObject, Spec, Time_old, target_series="GDPC1")

# Interactively zoom into different months to see which specific block
# drove the negative or positive signals associated with GDP!
fig.show()
```
